### Filter out only mixes with flyash as SCM

```python
import pandas as pd

# Load in cleaned data
df = pd.read_pickle('../02_processed_data/epd_data_cleaned_all.pkl')
df.head()
```

```python
#filter for mixes where there is a value for cementitious.fly_ash
#df_fly_ash = df[df['cementitious.fly_ash'].notna()]

#filter for mixes where there is a value for cementitious.fly_ash or cementitious.ggbs
df_fly_ash_or_ggbs = df[(df['cementitious.fly_ash'].notna()) | (df['cementitious.ggbs'].notna())]

# report the number of records with fly ash or ggbs
print(f"Number of records with fly ash or ggbs: {len(df_fly_ash_or_ggbs)}")
```

```python
# Save the filtered dataframe to a csv in my processed data folder
#df_fly_ash.to_csv('../02_processed_data/epd_data_fly_ash.csv', index=False)
df_fly_ash_or_ggbs.to_csv('../02_processed_data/epd_data_fly_ash_or_ggbs.csv', index=False)
```
